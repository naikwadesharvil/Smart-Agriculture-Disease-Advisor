```python
# 🌿 Smart Agriculture Disease Advisor

## Disease Knowledge Base

### Objective

This notebook creates a comprehensive knowledge base for all plant diseases in the PlantVillage dataset. After the CNN predicts a disease, the system retrieves detailed information including crop name, disease description, symptoms, causes, treatment, and prevention recommendations.

### Goals

- Create a disease information database
- Store treatment recommendations
- Store prevention measures
- Retrieve disease information automatically
- Integrate with the prediction system
```

```python
import pandas as pd
import json
```

```python
disease_database = {

    "Peach___Bacterial_spot": {

        "Crop": "Peach",

        "Disease": "Bacterial Spot",

        "Description": "A bacterial disease that causes dark spots on peach leaves and fruits.",

        "Symptoms": [
            "Brown spots on leaves",
            "Yellow halos around lesions",
            "Premature leaf drop"
        ],

        "Causes": [
            "Xanthomonas bacteria",
            "Warm and humid weather"
        ],

        "Treatment": [
            "Apply copper-based bactericides",
            "Remove infected leaves"
        ],

        "Prevention": [
            "Use disease-free plants",
            "Avoid overhead irrigation",
            "Improve air circulation"
        ],

        "Severity": "Medium",

        "Organic_Treatment": [
            "Neem oil spray",
            "Compost tea"
        ],

        "Recommended_Chemicals": [
            "Copper Oxychloride",
            "Copper Hydroxide"
        ]
    },

    "Apple___Apple_scab": {

        "Crop": "Apple",

        "Disease": "Apple Scab",

        "Description": "A fungal disease that affects apple leaves and fruits.",

        "Symptoms": [
            "Olive-green spots",
            "Dark lesions",
            "Leaf curling"
        ],

        "Causes": [
            "Venturia inaequalis fungus"
        ],

        "Treatment": [
            "Apply fungicide",
            "Remove infected leaves"
        ],

        "Prevention": [
            "Plant resistant varieties",
            "Maintain orchard hygiene"
        ],

        "Severity": "High",

        "Organic_Treatment": [
            "Sulfur spray"
        ],

        "Recommended_Chemicals": [
            "Captan",
            "Mancozeb"
        ]
    }

}
```

```python
with open("disease_database.json", "w") as file:
    json.dump(disease_database, file, indent=4)

print("✅ Disease database saved successfully.")
```

```python
with open("disease_database.json", "r") as file:
    disease_database = json.load(file)

print("Database loaded successfully.")
```

```python
print("Available Diseases:\n")

for disease in disease_database.keys():
    print("-", disease)
```

```python
predicted_disease = "Peach___Bacterial_spot"

info = disease_database[predicted_disease]

print("=" * 50)

print("Crop:", info["Crop"])
print("Disease:", info["Disease"])
print("Description:", info["Description"])

print("\nSymptoms")
for item in info["Symptoms"]:
    print("•", item)

print("\nCauses")
for item in info["Causes"]:
    print("•", item)

print("\nTreatment")
for item in info["Treatment"]:
    print("•", item)

print("\nPrevention")
for item in info["Prevention"]:
    print("•", item)

print("\nSeverity:", info["Severity"])

print("\nOrganic Treatment")
for item in info["Organic_Treatment"]:
    print("•", item)

print("\nRecommended Chemicals")
for item in info["Recommended_Chemicals"]:
    print("•", item)
```

```python
df = pd.DataFrame.from_dict(disease_database, orient="index")

df
```

```python
df.to_csv("disease_database.csv", index=True)

print("✅ CSV exported successfully.")
```
